```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
xticks = np.linspace(0, 360, 9)
```

## 2025-12-29

All azimuths are measured from north through east. We assume the northernmost radial support is at an azimuth of zero.

A is the distance from the inner edge of the interface ring to the inner edge of the radial support plate

B is the distance from a fixed point on the radial support column to the radial track

A and B are measured at each radial support after rotating the dome so that the center of the slit is aligned with the radial support.

C is distance A measured at the northern radial support while rotating the dome.

There are 10 radial supports. We duplicate the first one at azimuths of 0 and 360 degrees.


```python
A = np.array([106, 96, 84, 83, 93, 86, 93, 103, 106, 125, 106])
B = np.array([101, 108, 100, 110, 118, 107, 106, 113, 110, 101, 101])
AB = A + B
C = np.array([103, 104, 110, 108, 101, 101, 107, 105, 106, 102, 103])

thetaABC = np.linspace(0, 360, 11)

plt.plot(thetaABC, AB - np.mean(AB), color="C0", label="AB data")
plt.plot(thetaABC, C - np.mean(C), color="C1", label="C")

print((np.max(C) - np.min(C)) / 2)

theta = np.linspace(0, 360, 361)
plt.plot(theta, 15 * np.cos(np.radians(theta - 300)), color="C0", linestyle="dashed", label="AB model")
plt.legend()
plt.xlim(0,360)
plt.xticks(xticks)
plt.show()
```

This plot is consistent with a circular dome (with an error in radios of +/- 5 mm) being decentered by 15 mm towards an azimuth of 300.

D is the distance between the outer surface of the beams and the inner surface of the dome. It is measured every 1/16 of a rotation, but we are not able to measure it over the entrance doors, so two points are missing.

```python
D1 = np.array([30, 27, 32, 18, 14, 15, 26, 25, 30, 37, 43, 30, 42, 47, 30])
thetaD1 = np.array([0, 1, 2, 3, 4, 5, 8, 9, 10, 11, 12, 13, 14, 15, 16]) * 360 / 16

plt.plot(thetaD1, D1, color="C0", label="D1 data")

plt.plot(theta, 15 * np.cos(np.radians(theta - 300)) + np.mean(D1), color="C0", linestyle="dashed", label="D model")
plt.legend()
plt.xlim(0,360)
plt.xticks(xticks)
plt.ylabel("$D$ (mm)")
plt.xlabel("Azimuth (deg)")
plt.savefig("plot-D1.pdf")
plt.show()
```

This plot is consistent with a circular dome being decentered by 15 mm towards an azimuth of 300.

E and F are measurements of the distance from the building to the inner edge of the dome at a fixed point while rotating the dome.

E is from the outer edge of the beams.

F is from the outer edge of the wall panels

```python
E = np.array([31, 29, 39, 30, 33, 36, 27, 33, 31, 32, 31])
F = np.array([105, 98, 114, 98, 91, 106, 100, 110, 106, 100, 105])

thetaEF = np.linspace(0, 360, 11)
print((np.max(E) - np.min(E)) / 2)

plt.plot(thetaEF, E - np.mean(E), color="C0", label="E data")
plt.plot(thetaEF, F - np.mean(F), color="C1", label="F data")
plt.legend()
plt.xlim(0,360)
plt.xticks(xticks)
plt.show()
```

## 2025-12-30

We repeated the D measurement, as we wondered if there were errors in the previous measurement around 300 degrees. Both sets of measurements are consistent to (-7,+5) mm.

```python
D2 = [ 34, 31, 30, 25, 10, 18, 28, 25, 27, 32, 45, 31, 37, 50, 34]
thetaD2 = np.array([0, 1, 2, 3, 4, 5, 8, 9, 10, 11, 12, 13, 14, 15, 16]) * 360 / 16

theta = np.linspace(0,360,361)
Dmodel = 15 * np.cos(np.radians(theta - 300)) + np.mean(D1)

plt.plot(thetaD1, D1, color="C0", label="D1 data")
plt.plot(thetaD2, D2, color="C1", label="D2 data")
plt.plot(theta, Dmodel, color="C2", linestyle="dashed", label="D model")
plt.legend()
plt.xlim(0,360)
plt.ylim(0,60)
plt.xticks(xticks)
plt.ylabel("$D$ (mm)")
plt.xlabel("Azimuth (deg)")
plt.savefig("plot-D.pdf")
plt.show()

Dmodel=0
print(np.max([D1,D2])-np.min([D1,D2]))

Dmodel = 15 * np.cos(np.radians(thetaD1 - 300))

plt.plot(thetaD1, D1 - Dmodel, color="C0", label="D1 data")
plt.plot(thetaD2, D2 - Dmodel, color="C1", label="D2 data")
plt.axhline(np.mean(D1), color="C2", linestyle="dashed", label="D model")
plt.legend()
plt.xlim(0,360)
plt.ylim(0,60)
plt.xticks(xticks)
plt.ylabel(r"$D'$ (mm)")
plt.xlabel("Azimuth (deg)")
plt.savefig("plot-D-prime.pdf")
plt.show()

print(np.max([D1-Dmodel,D2-Dmodel])-np.min([D1-Dmodel,D2-Dmodel]))

for _thetaD1, _D1, _D2 in zip(thetaD1, D1, D2):
    print(r"%.1f&%.0f&%.0f\\" % (_thetaD1, _D1, _D2))

```
