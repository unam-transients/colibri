# Zero Points

- Images: 20251013T091001C1o onwards.
- Star: 00 00 02.16 +45 03 01.1

```python
import numpy as np
```

```python
filter = [ "g", "r", "i", "gri", "B", "z", "y", "zy"]
stdmag = np.array([ 14.57, 13.88, 13.60, 13.88, 14.57, 13.47, 13.40, 13.47, ])
dn = np.array([  205160, 371082, 316064, 871705, 68053, 176251, 87547, 257986 ])
gain = 2.23
exptime = 60
zp = dn * gain / exptime * np.pow(10, 0.4 * stdmag)
zpmodel = np.array([5.3e9, 5.0e9, 3.4e9, 1.3e10, 2.5e9, 1.5e9, 7.4e8, 2.2e9 ])

for filter, zp, zpmodel in zip(filter, zp, zpmodel):
    print("%s\t%.2e\t%.2f" % (filter, zp, zp / zpmodel))
```

The ZP model are from the observing manual. This confirms they are more or less correct, except for B.
