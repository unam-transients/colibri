# Zero Points

- Images: 20251013T091001C1o onwards.
- Star: 00 00 02.16 +45 03 01.1
- Photometry in grizy from PS via Vizier.
  - g = 14.57
  - r = 13.88
  - i = 13.60
  - z = 13.47
  - y = 13.40
- Photomtry in B estimated using the [SDSS transformation](https://classic.sdss.org/dr5/algorithms/sdssUBVRITransform.php) of Jester et al.:
  - `R-I = 1.02 * (13.88-13.60) + 0.21 = 0.50`
  - `B = 14.47 + 0.33 * (14.57-13.88) + 0.20 = 15.00`

```python
import numpy as np
```

```python
filter = [ "g", "r", "i", "gri", "B", "z", "y", "zy"]
stdmag = np.array([ 14.57, 13.88, 13.60, 13.88, 15.00, 13.47, 13.40, 13.47, ])
dn = np.array([  205160, 371082, 316064, 871705, 68053, 176251, 87547, 257986 ])
gain = 2.23
exptime = 60
zp = dn * gain / exptime * np.pow(10, 0.4 * stdmag)
zpmanual = np.array([5.3e9, 5.0e9, 3.4e9, 1.3e10, 2.5e9, 1.5e9, 7.4e8, 2.2e9 ])

for filter, zp, zpmanual in zip(filter, zp, zpmanual):
    print("%s\t%.2e\t%.2f" % (filter, zp, zp / zpmanual))
```

The ZP manual values are from the 12 Sep 2005 manual. This confirms they are more or less correct.
